# Ejemplo de agrupamiento jerárquico aglomerativo con el dataset Wine

## Objetivos

En este notebook utilizaremos el dataset Wine de `scikit-learn` para:

1. cargar y explorar brevemente los datos;
2. estandarizar las variables antes de calcular distancias;
3. aplicar agrupamiento jerárquico aglomerativo;
4. comparar los criterios de enlace single, complete, average y Ward;
5. construir e interpretar dendrogramas;
6. utilizar una línea de corte para obtener una partición con \(k\) clusters;
7. comparar las particiones mediante SSE/Inercia, Silhouette, Davies-Bouldin y Calinski-Harabasz;
8. visualizar e interpretar los clusters encontrados;
9. comparar al final con las clases reales del dataset como evaluación externa;
10. aplicar PCA y LLE como métodos de reducción de dimensionalidad;
11. comparar cuatro flujos que combinan reducción de dimensionalidad y clustering;
12. construir un ranking agregado mediante tres métricas internas.


## 1. Importación de librerías

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import LocallyLinearEmbedding
from sklearn.cluster import AgglomerativeClustering, KMeans
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score,
    adjusted_rand_score,
    normalized_mutual_info_score
)

from scipy.cluster.hierarchy import linkage, dendrogram, fcluster

pd.set_option("display.max_columns", None)
```

## 2. Carga del dataset

`load_wine()` contiene 178 observaciones, 13 características químicas numéricas y tres clases reales correspondientes a diferentes tipos de uvas.

Durante el clustering trabajaremos únicamente con \(X\).

```python
wine = load_wine()

X = pd.DataFrame(wine.data, columns=wine.feature_names)
y = pd.Series(wine.target, name="target")

print("Dimensiones de X:", X.shape)
print("Número de clases reales:", y.nunique())
X.head()
```

## 3. Exploración inicial

Antes de aplicar clustering jerárquico revisaremos:

- valores faltantes;
- estadísticos descriptivos;
- diferencias de escala entre variables.

Esto es especialmente importante porque los métodos jerárquicos trabajan con distancias entre observaciones o clusters.

```python
print("Valores faltantes por variable:")
display(X.isna().sum().to_frame("faltantes"))

print("\nEstadísticos descriptivos:")
display(X.describe().T)
```

### 3.1. Comparación de escalas

```python
ranges = pd.DataFrame({
    "min": X.min(),
    "max": X.max(),
    "rango": X.max() - X.min(),
    "media": X.mean(),
    "std": X.std()
}).sort_values("rango", ascending=False)

ranges
```

Las variables tienen escalas muy diferentes. Si calculáramos directamente las distancias euclidianas, las variables con rangos mayores podrían dominar el resultado.

Por esta razón aplicaremos estandarización.

## 4. Preprocesamiento

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)
X_scaled_df.describe().T[["mean", "std", "min", "max"]]
```

Después de `StandardScaler`, cada variable queda aproximadamente centrada en 0 y con desviación estándar cercana a 1.

## 5. Agrupamiento jerárquico aglomerativo

El algoritmo comienza con una observación por cluster:

En cada iteración:

1. calcula o actualiza las distancias entre los clusters actuales;
2. identifica el par de clusters más cercano según el criterio de enlace;
3. fusiona ese par;
4. repite hasta obtener un único cluster.

La principal decisión es cómo definir la distancia entre dos clusters. Compararemos:

- `single`: mínima distancia entre pares de observaciones;
- `complete`: máxima distancia entre pares;
- `average`: promedio de todas las distancias entre pares;
- `ward`: fusión que produce el menor incremento en la variabilidad intra-cluster (SSE).

## 6. Construcción de los dendrogramas

```python
linkage_methods = ["single", "complete", "average", "ward"]

Z_dict = {
    method: linkage(X_scaled, method=method, metric="euclidean")
    for method in linkage_methods
}

for method, Z in Z_dict.items():
    print(f"{method:>8}: matriz linkage con forma {Z.shape}")
```

La matriz de `linkage` almacena toda la historia de fusiones. Cada fila representa una fusión e incluye:

$$
[\text{cluster 1},\ \text{cluster 2},\ \text{distancia de fusión},\ \text{número de observaciones}]
$$

La altura de cada unión en el dendrograma corresponde a la distancia o disimilitud a la cual ocurrió la fusión.

### 6.1. Línea de corte para obtener \(k=3\) clusters

Para mostrar cómo se transforma un dendrograma en una partición, comenzaremos con \(k=3\). Una línea horizontal colocada entre las dos alturas de fusión apropiadas corta el dendrograma en tres ramas principales.

Aquí _k=3_ se utiliza primero con propósito ilustrativo. Más adelante se utilizan las métricas internas para estudiar si este valor está bien respaldado.

```python
def cut_height_for_k(Z, k):
    """Altura intermedia que produce k clusters al cortar el dendrograma."""
    n = Z.shape[0] + 1
    if k <= 1 or k >= n:
        raise ValueError("k debe estar entre 2 y n-1.")
    lower = Z[n - k - 1, 2]
    upper = Z[n - k, 2]
    return (lower + upper) / 2

k_cut = 3
cut_heights = {
    method: cut_height_for_k(Z, k_cut)
    for method, Z in Z_dict.items()
}

cut_heights
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

for ax, method in zip(axes.ravel(), linkage_methods):
    Z = Z_dict[method]
    h = cut_heights[method]

    dendrogram(
        Z,
        ax=ax,
        no_labels=True,
        color_threshold=h,
        above_threshold_color="gray"
    )
    ax.axhline(h, linestyle="--", linewidth=2, label=f"Corte para k={k_cut}")
    ax.set_title(f"Linkage: {method}")
    ax.set_xlabel("Observaciones")
    ax.set_ylabel("Distancia / disimilitud")
    ax.legend()

plt.tight_layout()
plt.show()
```

### ¿Cómo interpretar la línea de corte?

La regla práctica es: el número de ramas verticales que atraviesa la línea horizontal corresponde al número de clusters obtenidos.
Por tanto, al mover la línea de corte hacia arriba se obtienen menos clusters; al moverla hacia abajo se obtienen más clusters.

## 7. Obtener las particiones para cada criterio de enlace

```python
labels_k3 = {}

for method, Z in Z_dict.items():
    # fcluster etiqueta desde 1; restamos 1 para trabajar con 0,1,2,...
    labels = fcluster(Z, t=k_cut, criterion="maxclust") - 1
    labels_k3[method] = labels

    counts = pd.Series(labels).value_counts().sort_index()
    print(f"\n{method.upper()} - tamaños de los clusters")
    print(counts)
```

## 8. Evaluación interna de las particiones

Utilizaremos cuatro métricas:

- Inercia: suma de cuadrados de las distancias de cada observación al centroide de su cluster. Menor es mejor para una partición con el mismo _k_.
- Silhouette: combina cohesión y separación. Mayor es mejor.
- Davies-Bouldin: compara dispersión interna y separación entre clusters. Menor es mejor.
- Calinski-Harabasz: compara separación entre clusters con dispersión interna. Mayor es mejor.

```python
def clustering_sse(X_array, labels):
    sse = 0.0
    for cluster_id in np.unique(labels):
        points = X_array[labels == cluster_id]
        centroid = points.mean(axis=0)
        sse += np.sum((points - centroid) ** 2)
    return sse

def internal_metrics(X_array, labels):
    return {
        "SSE ↓": clustering_sse(X_array, labels),
        "Silhouette ↑": silhouette_score(X_array, labels),
        "Davies-Bouldin ↓": davies_bouldin_score(X_array, labels),
        "Calinski-Harabasz ↑": calinski_harabasz_score(X_array, labels),
    }

rows = []
for method, labels in labels_k3.items():
    row = {"Linkage": method}
    row.update(internal_metrics(X_scaled, labels))
    rows.append(row)

metrics_k3 = pd.DataFrame(rows).set_index("Linkage")
metrics_k3.round(3)
```

## 9. ¿Qué ocurre si cambiamos el número de clusters?

Ahora evaluaremos: $$[k=2,3,\ldots,8]$$ para cada criterio de enlace, lo que permitirá estudiar conjuntamente: el criterio de enlace, y el nivel de corte del dendrograma.

```python
k_values = range(2, 9)
results = []

for method, Z in Z_dict.items():
    for k in k_values:
        labels = fcluster(Z, t=k, criterion="maxclust") - 1

        row = {
            "linkage": method,
            "k": k,
        }
        row.update(internal_metrics(X_scaled, labels))
        results.append(row)

results = pd.DataFrame(results)
results.round(3)
```

### 9.1. Métricas frente al número de clusters

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

for method in linkage_methods:
    subset = results[results["linkage"] == method]

    axes[0, 0].plot(subset["k"], subset["SSE ↓"], marker="o", label=method)
    axes[0, 1].plot(subset["k"], subset["Silhouette ↑"], marker="o", label=method)
    axes[1, 0].plot(subset["k"], subset["Davies-Bouldin ↓"], marker="o", label=method)
    axes[1, 1].plot(subset["k"], subset["Calinski-Harabasz ↑"], marker="o", label=method)

axes[0, 0].set_title("SSE / Inercia (menor es mejor)")
axes[0, 1].set_title("Silhouette (mayor es mejor)")
axes[1, 0].set_title("Davies-Bouldin (menor es mejor)")
axes[1, 1].set_title("Calinski-Harabasz (mayor es mejor)")

for ax in axes.ravel():
    ax.set_xlabel("Número de clusters (k)")
    ax.grid(alpha=0.2)
    ax.legend()

plt.tight_layout()
plt.show()
```

### 9.2. Mejor configuración según cada métrica

```python
best_sil = results.loc[results["Silhouette ↑"].idxmax()]
best_db = results.loc[results["Davies-Bouldin ↓"].idxmin()]
best_ch = results.loc[results["Calinski-Harabasz ↑"].idxmax()]

best_summary = pd.DataFrame({
    "Métrica": ["Silhouette", "Davies-Bouldin", "Calinski-Harabasz"],
    "Criterio": ["máximo", "mínimo", "máximo"],
    "Linkage": [best_sil["linkage"], best_db["linkage"], best_ch["linkage"]],
    "k": [int(best_sil["k"]), int(best_db["k"]), int(best_ch["k"])],
    "Valor": [
        best_sil["Silhouette ↑"],
        best_db["Davies-Bouldin ↓"],
        best_ch["Calinski-Harabasz ↑"]
    ]
})

best_summary.round(3)
```

Nota: SSE no se usa aquí para escoger simplemente su mínimo global, porque al aumentar _k_ la SSE tiende naturalmente a disminuir. Su comportamiento debe analizarse en conjunto con las demás métricas y con el dendrograma.

# 13. Comparación de cuatro flujos de trabajo

En la sección anterior PCA se utilizó solamente para proyectar en dos dimensiones una partición calculada con las 13 variables estandarizadas.

En esta sección PCA y LLE forman parte del flujo de modelado. Ambos métodos reducen los datos de 13 a 8 dimensiones y los algoritmos de clustering trabajan con las ocho dimensiones resultantes.

Se compararán estas combinaciones:

- K-Means + PCA
- K-Means + LLE
- agrupamiento jerárquico con Ward + PCA
- agrupamiento jerárquico con Ward + LLE

En todos los casos se utiliza k = 3 para mantener constante el número de clusters. Las clases reales del dataset no participan en la construcción de las particiones.


## 13.1. Reducción de dimensionalidad de 13 a 8 dimensiones

Para este ejemplo se preestablece una reducción de 13 a 8 dimensiones. La finalidad es disminuir la dimensionalidad antes del clustering sin restringir el análisis a un plano bidimensional.

PCA construye componentes lineales ortogonales ordenados según la varianza que explican. Con ocho componentes se conserva aproximadamente el 92 % de la varianza total de las variables estandarizadas.

LLE construye una representación no lineal intentando conservar las relaciones de vecindad local. En este caso se generan también ocho dimensiones y se utilizan 12 vecinos.

LLE no proporciona una razón de varianza explicada equivalente a la de PCA. Por ello, el número de dimensiones y el número de vecinos deben analizarse mediante la calidad y estabilidad de los resultados obtenidos.

El valor de ocho se trata aquí como una decisión didáctica preestablecida. En una aplicación real convendría comparar varios números de componentes, varios valores de vecinos y la estabilidad de las particiones.


```python
n_components_reduction = 8
n_neighbors_lle = 12

pca_comparison = PCA(n_components=n_components_reduction)
X_pca_comparison = pca_comparison.fit_transform(X_scaled)

lle_comparison = LocallyLinearEmbedding(
    n_components=n_components_reduction,
    n_neighbors=n_neighbors_lle,
    method="standard",
    eigen_solver="arpack",
    random_state=42
)
X_lle_comparison = lle_comparison.fit_transform(X_scaled)

representations = {
    "PCA": X_pca_comparison,
    "LLE": X_lle_comparison
}

print("Dimensión original:", X_scaled.shape)
print("Dimensión después de PCA:", X_pca_comparison.shape)
print("Dimensión después de LLE:", X_lle_comparison.shape)
print(
    "Varianza explicada acumulada por PCA:",
    round(pca_comparison.explained_variance_ratio_.sum(), 3)
)
```

### 13.1.1. Varianza explicada acumulada por PCA


```python
pca_full = PCA().fit(X_scaled)
cumulative_variance = np.cumsum(pca_full.explained_variance_ratio_)
components = np.arange(1, X_scaled.shape[1] + 1)

fig, ax = plt.subplots(figsize=(8.5, 4.8))
ax.plot(components, cumulative_variance, marker="o", color="#3264a8")
ax.axvline(n_components_reduction, color="#d95f0e", linestyle="--")
ax.axhline(
    cumulative_variance[n_components_reduction - 1],
    color="#777777",
    linestyle=":"
)
ax.scatter(
    [n_components_reduction],
    [cumulative_variance[n_components_reduction - 1]],
    color="#d95f0e",
    s=70,
    zorder=3
)
ax.annotate(
    f"8 componentes: {cumulative_variance[7]:.1%}",
    (8, cumulative_variance[7]),
    xytext=(8.4, cumulative_variance[7] - 0.12)
)
ax.set_xticks(components)
ax.set_ylim(0, 1.05)
ax.set_xlabel("Número de componentes")
ax.set_ylabel("Varianza explicada acumulada")
ax.set_title("Varianza explicada acumulada por PCA")
ax.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

## 13.2. Ejecución de los cuatro flujos

K-Means utiliza 20 inicializaciones y una semilla fija para favorecer la reproducibilidad.

El agrupamiento jerárquico utiliza Ward. Este criterio busca limitar el aumento de la variación interna producido por cada fusión y trabaja con distancia euclidiana.

Los dos algoritmos reciben como entrada las ocho dimensiones producidas por PCA o LLE.


```python
n_clusters = 3
flow_order = [
    "K-Means + PCA",
    "K-Means + LLE",
    "Jerárquico + PCA",
    "Jerárquico + LLE"
]

labels_flows = {}
evaluation_rows = []
cluster_size_rows = []

for reduction_name, X_reduced in representations.items():
    models = {
        "K-Means": KMeans(
            n_clusters=n_clusters,
            n_init=20,
            random_state=42
        ),
        "Jerárquico": AgglomerativeClustering(
            n_clusters=n_clusters,
            linkage="ward"
        )
    }

    for algorithm_name, model in models.items():
        flow_name = f"{algorithm_name} + {reduction_name}"
        labels = model.fit_predict(X_reduced)
        labels_flows[flow_name] = labels

        evaluation_rows.append({
            "Flujo": flow_name,
            "Silhouette ↑": silhouette_score(X_scaled, labels),
            "Davies–Bouldin ↓": davies_bouldin_score(X_scaled, labels),
            "Calinski–Harabasz ↑": calinski_harabasz_score(X_scaled, labels),
            "Silhouette reducido ↑": silhouette_score(X_reduced, labels),
            "Davies–Bouldin reducido ↓": davies_bouldin_score(X_reduced, labels),
            "Calinski–Harabasz reducido ↑": calinski_harabasz_score(
                X_reduced, labels
            )
        })

        counts = pd.Series(labels).value_counts().sort_index()
        cluster_size_rows.append({
            "Flujo": flow_name,
            "Cluster 0": counts.get(0, 0),
            "Cluster 1": counts.get(1, 0),
            "Cluster 2": counts.get(2, 0)
        })

comparison = (
    pd.DataFrame(evaluation_rows)
    .set_index("Flujo")
    .loc[flow_order]
)

cluster_sizes = (
    pd.DataFrame(cluster_size_rows)
    .set_index("Flujo")
    .loc[flow_order]
)
```

## 13.3. Tabla comparativa principal

Cada algoritmo construye los clusters en su representación reducida de ocho dimensiones. Sin embargo, para comparar los cuatro resultados sobre una base común, las etiquetas obtenidas se evalúan con las 13 variables originales estandarizadas.

Esta tabla se utiliza para elaborar el ranking principal.


```python
#13 variables originales estandarizadas
common_space_metrics = comparison[[
    "Silhouette ↑",
    "Davies–Bouldin ↓",
    "Calinski–Harabasz ↑"
]]

common_space_metrics.round(3)
```

## 13.4. Métricas en el espacio reducido

La siguiente tabla evalúa cada partición en las ocho dimensiones donde se ejecutó el clustering.

Estos resultados ayudan a estudiar la estructura obtenida dentro de cada representación, pero PCA y LLE producen geometrías diferentes. Por esa razón, esta tabla se considera complementaria y no se utiliza para el ranking principal.


```python
#Ejemplos reducidos a 8 dimensiones

reduced_space_metrics = comparison[[
    "Silhouette reducido ↑",
    "Davies–Bouldin reducido ↓",
    "Calinski–Harabasz reducido ↑"
]]

reduced_space_metrics.round(3)
```

## 13.5. Ranking agregado

Las tres métricas tienen escalas distintas, por lo que sus valores no se suman directamente.

Se asigna a cada flujo un rango entre 1 y 4:

- Silhouette: el mayor valor recibe rango 1.
- Davies–Bouldin: el menor valor recibe rango 1.
- Calinski–Harabasz: el mayor valor recibe rango 1.

La suma más baja ocupa la primera posición.

Este ranking es una regla de síntesis didáctica. No representa una prueba estadística ni garantiza la existencia de una única partición correcta.


```python
ranking = pd.DataFrame(index=common_space_metrics.index)

ranking["Rango Silhouette"] = common_space_metrics["Silhouette ↑"].rank(
    ascending=False,
    method="min"
)
ranking["Rango Davies–Bouldin"] = common_space_metrics[
    "Davies–Bouldin ↓"
].rank(
    ascending=True,
    method="min"
)
ranking["Rango Calinski–Harabasz"] = common_space_metrics[
    "Calinski–Harabasz ↑"
].rank(
    ascending=False,
    method="min"
)

ranking["Suma de rangos ↓"] = ranking.sum(axis=1)
ranking = ranking.sort_values([
    "Suma de rangos ↓",
    "Rango Silhouette",
    "Rango Davies–Bouldin"
])
ranking.insert(0, "Posición", range(1, len(ranking) + 1))
ranking = ranking.astype(int)

ranking
```

## 13.6. Comparación gráfica de las métricas


```python
colors = {
    "K-Means + PCA": "#3264a8",
    "K-Means + LLE": "#d95f0e",
    "Jerárquico + PCA": "#6baed6",
    "Jerárquico + LLE": "#fdae6b"
}

fig, axes = plt.subplots(1, 3, figsize=(17, 4.8))
metric_specs = [
    ("Silhouette ↑", "Silhouette\n(mayor es mejor)"),
    ("Davies–Bouldin ↓", "Davies–Bouldin\n(menor es mejor)"),
    ("Calinski–Harabasz ↑", "Calinski–Harabasz\n(mayor es mejor)")
]

for ax, (column, title) in zip(axes, metric_specs):
    values = common_space_metrics[column]
    bars = ax.bar(
        values.index,
        values.values,
        color=[colors[name] for name in values.index]
    )
    ax.set_title(title)
    ax.tick_params(axis="x", rotation=32)
    ax.grid(axis="y", alpha=0.25)
    ax.bar_label(bars, fmt="%.3f", padding=3, fontsize=9)

plt.suptitle("Comparación en el espacio original estandarizado", y=1.03)
plt.tight_layout()
plt.show()
```

## 13.7. Proyección bidimensional de las particiones

Los algoritmos fueron entrenados con las ocho dimensiones reducidas. Para construir estas gráficas se muestran solamente las dos primeras coordenadas de cada representación.

Por tanto, las figuras son una ayuda visual y no muestran toda la información utilizada para formar los clusters. Un solapamiento visible en el plano no implica necesariamente que los grupos se solapen en las ocho dimensiones.

Con la reducción a ocho dimensiones, los dos flujos basados en PCA ocupan las primeras posiciones del ranking. K-Means + PCA obtiene el primer lugar, seguido por agrupamiento jerárquico + PCA.

Los ocho componentes de PCA conservan aproximadamente el 92 % de la varianza de las variables estandarizadas. Además, las particiones producidas después de PCA mantienen mejores valores de las métricas internas cuando se evalúan sobre el espacio original.

En esta configuración, los flujos basados en LLE producen resultados menos favorables. Esto no permite concluir que PCA sea siempre superior a LLE. La conclusión se limita al dataset Wine y a los parámetros utilizados: ocho dimensiones, 12 vecinos para LLE, tres clusters, Ward y los parámetros indicados para K-Means.

También se observa que agrupamiento jerárquico + LLE genera un cluster considerablemente mayor que los otros dos. Este desbalance ayuda a interpretar sus métricas y muestra por qué no conviene seleccionar un flujo usando un único indicador.
